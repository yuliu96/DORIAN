## Run DORIAN

Set `is_vertical=True` for vertical stripes and `is_vertical=False` for horizontal stripes.

```python
from dorian import DORIAN
import tifffile

input_path = "data/example.tif"
output_path = "data/example_restored.tif"

model = DORIAN()

recon = model.train(is_vertical = True,
                    display = False,
                    angle_offset = [-3],
                    x = tifffile.imread(input_path),
                    display_angle_orientation = True,
                    save_path = output_path,
                    )

print(f"Saved restored image to: {output_path}")
```

## Visualize input and output

```python
import numpy as np
import tifffile
import matplotlib.pyplot as plt

inp = tifffile.imread(input_path)
out = tifffile.imread(output_path)

def first_plane(x):
    x = np.asarray(x)
    while x.ndim > 2:
        x = x[0]
    return x

inp2d = first_plane(inp)
out2d = first_plane(out)

plt.figure(figsize=(6, 6))
plt.imshow(inp2d, cmap="gray")
plt.title("Input")
plt.axis("off")
plt.show()

plt.figure(figsize=(6, 6))
plt.imshow(out2d, cmap="gray")
plt.title("DORIAN output")
plt.axis("off")
plt.show()
```
